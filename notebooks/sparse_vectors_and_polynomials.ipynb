# A gentle tour of sparse vectors, complex numbers, and polynomials

This notebook is a practical introduction. The goal is to see how the library behaves in the kind of situations a user might actually encounter: first with sparse vectors, then with a tiny algebra of complex numbers, and finally with polynomials.

## 1. Sparse vectors as simple combinations

A sparse vector is just a collection of basis elements with coefficients. You can think of it as a linear combination where most terms are zero.

```python
from algcom import SparseVector

v = SparseVector({"x": 2, "y": -1})
w = SparseVector({"z": 3, "x": 1})

print(v)
print(w)
print(v + w)
```

You can also add a scalar multiple and inspect a single coordinate.

```python
u = SparseVector({"x": 1, "y": 2})
print(2 * u)
print(u["x"])
print(u["z"])
```

## 2. From words to more structured objects

The library also lets you work with basis elements that are words, such as tuples of symbols.

```python
from algcom.tensor import TensorWord

word = TensorWord(("x", "y", "z"))
print(word)
print(word.rank)
```

## 3. A very small algebra: the complex numbers

Now we use the algebra machinery. We choose a basis with two elements: "1" and "i". The product is the usual multiplication of complex numbers.

```python
from fractions import Fraction
from algcom import Algebra

def complex_product(left, right):
    if left == "1" and right == "1":
        return {"1": 1}
    if left == "1" and right == "i":
        return {"i": 1}
    if left == "i" and right == "1":
        return {"i": 1}
    if left == "i" and right == "i":
        return {"1": -1}
    raise ValueError("unexpected basis element")

one = SparseVector({"1": 1})
algebra = Algebra.from_rule(complex_product, one=one)

i = SparseVector({"i": 1})
print(algebra.m(i, i))
print(algebra.m(one, i))
print(algebra.m(i, one))
```

## 4. Exponential and logarithm in this algebra

The algebra object also gives you a few standard operations. Here we see them on a simple element.

```python
exp_i = algebra.exponential(i, order=5)
log_one_plus_i = algebra.logarithm(one + i, order=5)

print(exp_i)
print(log_one_plus_i)
```

## 5. Polynomials

The polynomial class uses the same algebra idea, but now the basis elements are monomials.

```python
from algcom.polynomial import Polynomial

p = Polynomial({0: -1, 2: 1})
q = Polynomial({0: 1, 1: 1})

print(p)
print(q)
print(q.multiply(p))
print(p.evaluate_at(1))
```

That is the basic pattern: a small basis, a product rule, and then the generic algebra operations work for you.
